# Linear Algebra Implementation

```python
import numpy as np
import scipy as sc
from numpy.linalg import det
from numpy.linalg import inv
from numpy.linalg import eig
```

```python
# definition of vector & matrix
vector_a = np.array([2, 4, 6])
vector_b = np.array([1, 3, 5])
matrix_a = np.array([
[2, 1, 3],
[1, 0, 2],
[4, 1, 1]
])
matrix_b = np.array([
[1, 2, 0],
[3, 1, 4],
[2, 5, 1]
])
```

```python
# vector addition,sub,dot product, 

add = vector_a + vector_b
print(f"vector addition:\n {add}")

sub = vector_a - vector_b
print(f"vector subraction:\n {sub}")

dot_product = np.dot(vector_a,vector_b)
print(f"vector dotproduct:\n {dot_product}")
```

```python
# matrix multiply & add
add = matrix_a + matrix_b
print(f"matrix addition:\n {add}")

dot_product = np.dot(matrix_a,matrix_b)
print(f"vector dotproduct:\n {dot_product}")
```

```python
print(f"Matrix determinant:\n {det(matrix_a)}")
```

# Probability Concepts

```python
sales_success = [0.65, 0.35]
customer_segments = ["Premium", "Regular"]
segment_purchase_probability = {
"Premium": 0.80,
"Regular": 0.50
}
```

```python
# probability of successful sales

prob_success = sales_success[0]
print(prob_success)

premium_prob = segment_purchase_probability["Premium"]
print(f"P(Success | Premium) = {premium_prob:.0%}")
premium_prob1 = segment_purchase_probability["Regular"]
print(f"P(Success | Regular) = {premium_prob1:.0%}")
```

```python
# bayes' theorem 

# Probability Dataset for Customer Purchase Analysis
# P(A): Probability that a customer is Premium
P_A = 0.40
# P(B): Probability that a customer makes a purchase
P_B = 0.62
# P(B|A): Probability that a Premium customer makes a purchase
P_B_given_A = 0.80

P_A_given_B = (P_B_given_A * P_B) / (P_A)
print(P_A_given_B)
```

```python
# binomial distribution

from scipy.stats import binom

from scipy.stats import binom

n_trials = 20
success_prob = 0.65
k = 20

probability = binom.pmf(k, n_trials, success_prob)

print(f"Probability of exactly {k} successful sales: {probability:.4f}")
```

```python
normal_mean = 50
normal_std = 10
uniform_low = 0
uniform_high = 1

# Generate 1000 random values
normal_data = np.random.normal(normal_mean, normal_std, 1000)
uniform_data = np.random.uniform(uniform_low, uniform_high, 1000)

print("Normal Distribution Sample:")
print(normal_data[:10])

print("\nUniform Distribution Sample:")
print(uniform_data[:10])
```

```python
# data visualise

import matplotlib.pyplot as plt

plt.figure(figsize=(12, 5))

# Normal Distribution
plt.subplot(1, 2, 1)
plt.hist(normal_data, bins=30)
plt.title("Normal Distribution")

# Uniform Distribution
plt.subplot(1, 2, 2)
plt.hist(uniform_data, bins=30)
plt.title("Uniform Distribution")
plt.show()
```

# Descriptive statistics

```python
# data set
import pandas as pd
sales_data = [120, 135, 150, 145, 160, 155, 170, 165, 180, 175,
190, 200, 210, 195, 185, 175, 165, 155, 145, 135]

df = pd.DataFrame(sales_data, columns=["Sales"])
df
```

```python
# mean , median, mode

print("Mean:", df["Sales"].mean())
print("Median:", df["Sales"].median())
print("Mode:")
print(df["Sales"].mode())
```

```python
# range, variance, and standard deviation

print("Range:", df["Sales"].max() - df["Sales"].min())
print("Standard Deviation:", df["Sales"].std())
print("Variance:", df["Sales"].var())
```

```python
#skewness

print(f"Skewness: {df["Sales"].skew()}")
print(f"Kurtosis: {df["Sales"].kurt()}")
```

The sales dataset is symmetric, with a mean and median of 165 and a skewness value close to zero. The data is fairly balanced around the center, with no noticeable positive or negative skew. The negative kurtosis indicates a flatter-than-normal distribution with relatively few outliers
